# 🧑‍🍳 Barba Niko: A Soulful AI Companion for Meal Planning and Reflection


This Kaggle notebook is the Capstone Project for the 5-day Gen AI Intensive Course with Google.
It introduces "Barba Niko," a generative AI agent designed to support users through culturally rich, personalized meal planning,
offering wisdom, nourishment, and joy. Inspired by Greek traditions and the warmth of a barefoot monk.


Gen AI Capabilities Used:
1. Agents
2. Function Calling
3. Few-shot Prompting
4. Structured Output (JSON)

```python
## 🪔 Introduction

print("\u03a7\u03b1\u03b9\u03c1\u03b5\u03c4\u03b5 \u03c4\u03bf\u03bd Barba Niko \u2014 your warm-hearted AI kitchen companion.")
print("He will help you plan meals with care, tradition, and a touch of soul.")
```

```python
## 🧰 Setting Up Tools
from typing import List, Dict
import json
```

```python
## 🤖 Barba Niko Agent Setup

class BarbaNiko:
    def __init__(self, name="Barba Niko"):
        self.name = name

    def greet(self):
        return "Kalosorisate! What would you like to eat this week, my child?"

    def suggest_meals(self, preference: str) -> List[str]:
        meals = {
            "vegetarian": ["Gigantes plaki", "Spanakopita", "Briam"],
            "comfort": ["Pastitsio", "Youvarlakia avgolemono", "Moussaka"],
            "light": ["Ladera", "Grilled fish with horta", "Dakos"]
        }
        return meals.get(preference.lower(), ["Fasolada", "Gemista", "Lentil soup"])

    def bless_meal(self, meal: str) -> str:
        return f"Ahh, {meal}... A dish even the angels would wait in line for. Kali orexi!"

    def meal_plan_json(self, preference: str) -> str:
        meals = self.suggest_meals(preference)
        plan = {"BarbaNiko_WeeklyMealPlan": meals}
        return json.dumps(plan, indent=2)
```

```python
## 🍽️ Simulated Conversation with Barba Niko

barba = BarbaNiko()
print(barba.greet())

selected_meals = barba.suggest_meals("comfort")
print("\nBarba Niko suggests these comforting meals:")
for m in selected_meals:
    print(f"- {m}")

print("\nStructured meal plan:")
print(barba.meal_plan_json("comfort"))
```
